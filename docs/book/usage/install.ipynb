# Installation

This page covers installing `mboostrs` for Python and for R, the platforms with
prebuilt packages, and a check that the installation works. It is for anyone
setting up the package. Building from source is needed only to modify the Rust
core.

`mboostrs` is distributed as a compiled extension (a Rust core behind a Python
binding), so `pip` installs a prebuilt wheel, a precompiled package file. No Rust
toolchain is required for end users.

## Python

```bash
pip install mboostrs
```


Requirements: Python ≥ 3.11.

(wheel-availability)=
### Wheel availability

Wheels are built for CPython 3.11, 3.12 and 3.13.

| Platform     | x86_64              | aarch64 / ARM64     |
|--------------|---------------------|---------------------|
| Linux        | ✓ (manylinux2014)   | ✓ (manylinux2014)   |
| macOS        | ✓ (macOS 10.12+)    | ✓ (macOS 11+)       |
| Windows      | ✓                   | —                   |

Each release builds four wheel targets per Python version: one macOS
`universal2` wheel that contains both Apple architectures (the Intel part
requires macOS 10.12 or later, the Apple Silicon part macOS 11 or later),
manylinux2014 wheels for Linux x86_64 and Linux aarch64, and a Windows x86_64
wheel. Windows on ARM gets a source build.

Released wheels use the portable baseline for their platform. CPU-specific
instructions may be enabled for a local benchmark build, but are not used in
generally tagged wheels.

On other platforms and on Python versions without a wheel, pip falls back to a source build, which requires the Rust toolchain described
under Building from source.

The base package installs `numpy` and `polars`. Optional extras provide
plotting (`mboostrs[plot]`), pandas output (`mboostrs[pandas]`), Arrow
interchange (`mboostrs[arrow]`), and scikit-learn estimators
(`mboostrs[sklearn]`):

```bash
pip install "mboostrs[sklearn]"
```


## Building from source

To build the Rust core yourself, for example to develop `mboostrs`:

```bash
# Rust toolchain and maturin
curl https://sh.rustup.rs -sSf | sh
pip install maturin
# from the repository root
maturin develop --release
```


[maturin](https://www.maturin.rs/) is the build tool that compiles a Rust
library into a Python extension module and packages it as a wheel.
`maturin develop --release` compiles the Rust core (`crates/mboostrs-core`) and
the Python binding (`crates/mboostrs-py`, written with
[PyO3](https://pyo3.rs/), the Rust library for writing Python extension
modules), then installs the resulting extension into the active environment.

## R

The R package calls the same Rust core through
[`extendr`](https://extendr.github.io/), the Rust library for writing R
extensions. Installing it compiles the Rust code, so it needs the Rust
toolchain from Building from source. From the repository:

```r
# from a shell:
#   R CMD INSTALL mboostrs
#
# or from R (requires the devtools package):
devtools::install_local("mboostrs")
```

```{note}
The R binding exposes `glmboost`/`gamboost`/`glmboostLSS`/`gamboostLSS`,
`cvrisk`, AIC, hatvalues, confint, stability selection, variable importance,
plotting helpers, and the core family/base-learner workflows. A few advanced
diagnostics and prediction paths differ by language. Use
[](interface_map#querying-capabilities-from-code) for capability limits and runtime discovery, and
[](../appendix/r_parity_matrix) for the detailed parity matrix.
```

## Verifying the install

```python
import mboostrs as mb
print(mb.__version__)
```
