## Conventions

- Vectors are columns; $\hat y = X \hat\beta$ is a length-$n$ vector.
- Subscripts index observations ($y_i$), superscripts in brackets index
  boosting iterations ($\hat f^{[m]}$).
- A hat ($\hat{\cdot}$) over a quantity denotes an estimate (e.g. $\hat\beta$, $\hat f$).
- $\operatorname{tr}$, $\operatorname{rank}$, $\operatorname{diag}$ have
  their standard meanings. $\ker$ is the null space.
- "Pseudo-residual" and "negative gradient" are used interchangeably for $u$.
- "Link scale" and "predictor scale" both mean the scale of $f$ (before the
  inverse link is applied).
- $K$ is overloaded: the penalty matrix in "Base learners and penalization"
  and the number of distribution parameters in "Distributional (GAMLSS)
  models" are unrelated uses of the same letter.
