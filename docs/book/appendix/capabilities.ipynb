## Status vocabulary

- `verified`: compared with the named oracle at the recorded tolerance.
- `smoke`: executable behavior is covered, without a full external numerical oracle.
- `intentional-deviation`: tested behavior differs deliberately; the validation record links its rationale.
- `unsupported`: the frontend rejects or does not expose the operation.
- `unverified`: code may exist, but the manifest does not make a correctness claim.
