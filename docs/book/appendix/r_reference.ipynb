# R reference

Every exported function of the `mboostrs` R package that has a help page, with its call signature and the title of its help page, and the S3 generics the package extends. Type `?name` in R for the argument descriptions.

137 exported entries and 60 S3 generics. Internal helpers are not listed.

The R frontend reads a narrower formula language than the Python frontend. It refuses the `%X%` (tensor-product) and `%O%` (Kronecker-product) operators, which combine two base learners into one interaction learner; the `bins=` argument, which bins numeric covariates to speed up fitting; and inline transforms such as `bbs(log(x))`. Support for `by=` (a varying-coefficient modifier) depends on the learner and the type of the modifier; an unsupported combination stops with an error. The [migration guide from R mboost](../usage/r_migration.ipynb) gives the Python spelling of each refused feature.

## REML refits

A REML refit takes the terms a boosted model selected and refits them as one penalized additive model, with smoothing parameters chosen by restricted maximum likelihood (REML). The refit has a coefficient covariance matrix (`vcov`), from which standard errors and intervals follow.

| Call | What it is |
|---|---|
| `concurvity(object, pairwise = FALSE, ...)` | Concurvity of a model's terms |
| `gam_refit(model, terms = NULL, which = c("selected", "all"))` | Refit a boosted additive model with REML smoothing parameters |
| `gamlss_draws( object, newdata, n_draws = 1000, seed = 0, corrected = FALSE, unseen_levels = c("error", "zero_effect") )` | Posterior draws of a GAMLSS refit's distribution parameters |
| `gamlss_refit(model, terms = NULL, which = c("selected", "all"))` | Refit a distributional model with REML smoothing parameters |

## Prediction intervals

`conformal_interval` methods: `"split"` calibrates the interval on a held-out share of the training rows (one refit); `"cv+"` is the cross-validation-plus interval, which calibrates on the out-of-fold residuals of K refits so every row is used; `"cqr"` is conformalized quantile regression, which fits a lower and an upper quantile model and calibrates their band, so the width varies with the predictors; `"distributional"` calibrates the central band of a fitted distributional model.

| Call | What it is |
|---|---|
| `conformal_interval( model, newdata = NULL, level = 0.9, method = c("split", "cv+", "cqr", "distributional", "adaptive"), seed = NULL, calib_fraction = 0.5, n_folds = 10L, mstop = NULL, calibration = NULL, score = "auto", count_reference = "zero_inflated_poisson", decay = 0.99, time = NULL )` | Conformal prediction intervals for a boosted model Also documents `adaptive_conformal_stream`. |
