# Case studies

Eighty-one worked analyses demonstrate fitting, tuning, interpretation and
prediction with mboostrs. They include introductory workflows, specialised
applications, and experiments that examine the methods and the package itself.
Choose a starting point below; the full catalogue follows.

Benchmarks are a separate part of the book. The benchmark notebooks still
live in this directory, because open drafts edit those paths; they are indexed
from [](../benchmarks/overview.ipynb) and belong with the benchmarks. This
catalogue still lists both. The unemployment forecast is
[](../case_studies/unemployment_forecast.ipynb).

## Introductory workflows

Start with [](bodyfat.ipynb) if you know regression and are new to the package.
It introduces a direct fit, cross-validation, an effect plot and a prediction
table, then compares straight and smooth effects. Its bootstrap and curvature
investigations are marked as optional extensions.

Continue with [](motorcycle_variance.ipynb) to give a distribution's standard
deviation its own predictor and produce conditional centiles. The main workflow
ends with an interval plot; extensions examine coverage and numerical settings.

Each study-design card lists the response, units, resampling design and
interpretation limits.

## Reading and running a study

Read the data description, run the fit and inspect its output before changing
the specification. Each analysis states what its result supports and the
population or task to which it applies. A simulated study can also compare its
result with the generating process.

The downloadable notebooks contain the executable code, including the longer
extensions. Computed values in the revised introductory pages are inserted from
their outputs; interpretation still requires checking the executed page.
A successful build verifies execution and document structure, not the validity
of every statistical claim.

## Sample sizes

Some studies use small samples, including 71 women in the body-fat data and
46 recorded events in the survival example. Their comparisons can be sensitive
to the fold assignment and to which observations are present. The resampling
extensions examine that sensitivity; the main workflows explain how to obtain
and interpret the fitted model.
